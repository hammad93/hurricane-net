```python
import pandas
df = pandas.read_csv("https://raw.githubusercontent.com/hammad93/hurricane-net/refs/heads/main/docs/archive/server_sum_power.csv")
df.head()
```

```python
pwr_col = 'state' # kWh
time_col = 'last_changed' # ISO timestamp
df[time_col] = pandas.to_datetime(df[time_col])
sorted = df.sort_values(by=time_col)
total_kwh = sorted.iloc[-1][pwr_col] - sorted.iloc[0][pwr_col]
total_days = (
    sorted.iloc[-1][time_col] - sorted.iloc[0][time_col]).days
# average per day * total days in year
avg_kwh_yr = 365 * (total_kwh / total_days)
print(f"Summation power delivered (kWh) in {total_days} days: {total_kwh:.2f}")
print(f"Average monthly power consumption: {avg_kwh_yr/12:.2f} kWh")
print(f"Average yearly power consumption: {avg_kwh_yr:.2f} kWh")
```

```python
# https://www.epa.gov/egrid/summary-data
# https://www.epa.gov/system/files/documents/2025-06/summary_tables_rev2.pdf
# eGRID Subregion Total Output Emission Rates (lb/MWh)
# Based on location of the server in PDF map, the value for RFCE is 596.904
emission_rate = 596.904 # lb/MWh
# lb/MWh * (kWh/year)/1000
carbon_emissions_lbs = emission_rate * (avg_kwh_yr / 1000)
# tCO2 -> Metric Tonnes of Carbon Dioxide. 2200 lbs in 1 metric tonne
print(f"Equivalent Scope 2 CO2 emissions: {carbon_emissions_lbs:.2f} lbs/year ({carbon_emissions_lbs/2200:.2f} tCO2/year)")
```
